## Parte 1: Variables y operaciones básicas

```python
# Estado inicial
saldo = 500_000      # Saldo de la cuenta
deposito = 150_000   # Monto a depositar
retiro = 80_000      # Monto a retirar

# Actualización del estado
saldo = saldo + deposito   # Se deposita dinero
saldo = saldo - retiro     # Se retira dinero

# Salida del estado final
print("Saldo final de la cuenta:", saldo)
```

```python
# Declaramos nuevas variables
interes = 0.02  # 2% mensual
saldo_final = saldo * (1 + interes)

# Mostramos el resultado usando f-string
print(f"Saldo final con interés: ${saldo_final:.2f}")
```

```python
# Modificamos el código para que el usuario ingrese los valores
saldo = float(input("Ingrese el saldo inicial: "))
deposito = float(input("Ingrese el monto a depositar: "))
retiro = float(input("Ingrese el monto a retirar: "))

# Aplicamos interés
interes = 0.02
saldo_final = (saldo + deposito - retiro) * (1 + interes)

print(f"Saldo final con interés: ${saldo_final:.2f}")
```

### Explicación del estado inicial y final:

- **Estado inicial**: Representa los valores iniciales de las variables antes de ejecutar las operaciones. En memoria, cada variable ocupa un espacio con su valor asignado.
- **Estado final**: Representa los valores después de aplicar todas las operaciones. Cada instrucción modifica el contenido de la memoria:
  - `saldo = saldo + deposito` crea un nuevo valor y lo asigna a la variable `saldo`
  - `saldo = saldo - retiro` nuevamente modifica el valor en memoria
  - `saldo_final = saldo * (1 + interes)` crea una nueva variable con el cálculo final

## Parte 2: Listas y operaciones

```python
# Lista inicial de temperaturas en °C
temperaturas = [22.5, 23.0, 21.8, 24.3, 25.1]

# Operaciones básicas
print("Temperaturas:", temperaturas)
```

```python
# Agregamos una nueva temperatura al final
temperaturas.append(26.0)
print("Después de agregar 26.0:", temperaturas)
```

```python
# Eliminamos la primera temperatura
temperaturas.pop(0)
print("Después de eliminar la primera:", temperaturas)
```

```python
# Insertamos una temperatura en la segunda posición
temperaturas.insert(1, 23.7)
print("Después de insertar 23.7 en posición 1:", temperaturas)
```

```python
# Mostramos la lista ordenada de menor a mayor
temperaturas_ordenadas = sorted(temperaturas)
print("Temperaturas ordenadas:", temperaturas_ordenadas)
```

```python
# Calculamos la diferencia entre la temperatura más alta y la más baja
diferencia = max(temperaturas) - min(temperaturas)
print(f"Diferencia (máx - mín): {diferencia}")

# Calculamos el promedio
promedio = sum(temperaturas) / len(temperaturas)
print(f"Promedio: {promedio:.2f}")
```

```python
# Encontramos el índice de la temperatura 23.7
try:
    indice = temperaturas.index(23.7)
    print(f"Índice de 23.7: {indice}")
except ValueError:
    print("La temperatura 23.7 no está en la lista")
```

```python
# Mostramos las temperaturas de la 2da posición a la 4ta posición (índices 1 a 3)
print("Temperaturas de la 2da a 4ta posición:", temperaturas[1:4])
```

```python
# Mostramos las tres primeras temperaturas
print("Tres primeras temperaturas:", temperaturas[:3])
```

## Parte 3: Diccionarios y listas de diccionarios

```python
# Lista de productos en una tienda
inventario = [
    {"nombre": "Pan", "precio": 1500, "stock": 30},
    {"nombre": "Leche", "precio": 3500, "stock": 15},
    {"nombre": "Café", "precio": 8000, "stock": 10},
]
```

```python
# Mostramos el nombre del primer producto
print("Primer producto:", inventario[0]["nombre"])
```

```python
# Agregamos un nuevo producto al inventario
nuevo_producto = {"nombre": "Azúcar", "precio": 2000, "stock": 25}
inventario.append(nuevo_producto)
print("Inventario actualizado:", inventario)
```

```python
# Disminuimos el stock de "Leche" en 2 unidades (venta)
for producto in inventario:
    if producto["nombre"] == "Leche":
        producto["stock"] -= 2
        break
print("Después de la venta de Leche:", inventario)
```

```python
# Calculamos el valor total del inventario
valor_total = sum(p["precio"] * p["stock"] for p in inventario)
print(f"Valor total del inventario: ${valor_total}")
```

```python
# Mostramos los productos cuyo stock sea menor a 10
bajo_stock = [p for p in inventario if p["stock"] < 10]
print("Productos con stock < 10:", bajo_stock)
```

```python
# Aumentamos el precio de todos los productos en un 10%
for producto in inventario:
    producto["precio"] = round(producto["precio"] * 1.10, 2)
print("Inventario con precios aumentados:", inventario)
```

```python
# Buscamos un producto por nombre (ingresado por el usuario)
nombre_buscar = input("Ingrese el nombre del producto a buscar: ")
encontrado = None
for producto in inventario:
    if producto["nombre"].lower() == nombre_buscar.lower():
        encontrado = producto
        break

if encontrado:
    print(f"Producto encontrado: {encontrado}")
else:
    print("Producto no encontrado")
```

```python
# Eliminamos un producto del inventario si su stock es 0
inventario = [p for p in inventario if p["stock"] > 0]
print("Inventario después de eliminar productos sin stock:", inventario)
```

```python
# Ordenamos la lista por precio ascendente
inventario_ordenado = sorted(inventario, key=lambda p: p["precio"])
print("Inventario ordenado por precio:", inventario_ordenado)
```

## Parte 4: Tuplas

```python
# Tupla: coordenadas (x, y)
punto = (3, 7)
print("Punto:", punto)
```

```python
# Creamos una tupla rectangulo = (ancho, alto)
rectangulo = (10, 5)

# Calculamos su área y perímetro
ancho, alto = rectangulo
area = ancho * alto
perimetro = 2 * (ancho + alto)
print(f"Área: {area}, Perímetro: {perimetro}")
```

```python
# Intentamos cambiar el valor de ancho
try:
    rectangulo[0] = 20
except TypeError as e:
    print(f"Error: {e}")
    print("Las tuplas son inmutables, no se pueden modificar sus elementos")
```

```python
# Calculamos la distancia total recorrida
import math

posiciones = [(0, 0), (1, 2), (3, 4)]
distancia_total = 0

for i in range(len(posiciones) - 1):
    x1, y1 = posiciones[i]
    x2, y2 = posiciones[i + 1]
    distancia = math.sqrt((x2 - x1)**2 + (y2 - y1)**2)
    distancia_total += distancia

print(f"Distancia total recorrida: {distancia_total:.2f}")
```

## Parte 5: Diccionarios anidados

```python
biblioteca = {
    "libros": [
        {"titulo": "1984", "autor": "George Orwell", "prestado": False},
        {"titulo": "El Principito", "autor": "Antoine de Saint-Exupéry", "prestado": True},
    ],
    "usuarios": ["Ana", "Luis", "María"]
}
```

```python
# Mostramos cuántos libros están prestados y cuántos disponibles
prestados = sum(1 for libro in biblioteca["libros"] if libro["prestado"])
disponibles = len(biblioteca["libros"]) - prestados
print(f"Libros prestados: {prestados}, Disponibles: {disponibles}")
```

```python
# Agregamos un nuevo libro
nuevo_libro = {"titulo": "Cien años de soledad", "autor": "Gabriel García Márquez", "prestado": False}
biblioteca["libros"].append(nuevo_libro)
print("Biblioteca actualizada:", biblioteca)
```

```python
# Marcamos como prestado el libro "1984"
for libro in biblioteca["libros"]:
    if libro["titulo"] == "1984":
        libro["prestado"] = True
        break

# Mostramos el estado actualizado del sistema
print(biblioteca)
```
