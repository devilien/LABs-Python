## Parte 1: Condicionales

```python
# Código base
edad = int(input("Ingrese su edad: "))
tiene_licencia = input("¿Tiene licencia? (s/n): ")
no_esta_sancionado = input("¿Está sancionado? (s/n): ")

# Determinamos si el usuario puede conducir
if edad >= 18 and tiene_licencia.lower() == 's' and no_esta_sancionado.lower() == 'n':
    print("Puede conducir")
else:
    print("No puede conducir")
```

## Parte 2: Bucles for

```python
# Código base
for i in range(1, 6):
    print(f"Iteración número {i}")
```

```python
# Calculamos la suma de los primeros n números naturales
n = int(input("Ingrese n: "))
suma = 0
for i in range(1, n + 1):
    suma += i
print(f"Suma de los primeros {n} números: {suma}")
```

```python
# Imprimimos solo los números pares usando if
for i in range(1, 6):
    if i % 2 == 0:
        print(f"Iteración número {i} (par)")
```

```python
# Agregamos un acumulador y mostramos el resultado final
acumulador = 0
for i in range(1, 6):
    acumulador += i
print(f"Resultado final del acumulador: {acumulador}")
```

```python
# Encontramos las temperaturas mayores al promedio
temperaturas = [22.5, 23.0, 21.8, 24.3, 25.1]
promedio = sum(temperaturas) / len(temperaturas)
mayores_promedio = [t for t in temperaturas if t > promedio]
print(f"Temperaturas mayores al promedio ({promedio:.2f}): {mayores_promedio}")
```

## Parte 3: Bucles while

```python
# Código base
contador = 0
while contador < 5:
    print("Contador:", contador)
    contador += 1
```

```python
# Modificamos el ciclo para contar hacia atrás
contador = 5
while contador > 0:
    print("Contador:", contador)
    contador -= 1
```

```python
# Programa que solicita números hasta que el usuario ingrese 0
numeros = []
while True:
    numero = float(input("Ingrese un número (0 para salir): "))
    if numero == 0:
        break
    numeros.append(numero)
print(f"Números ingresados: {numeros}")
```

### Explicación del estado del ciclo:

El estado del ciclo se representa por la variable contador. Esta variable controla la condición de continuación del bucle. En cada iteración, el estado cambia, lo que eventualmente hará que la condición del while deje de cumplirse y el bucle termine.

## Parte 4: Acumuladores pares e impares

```python
# Código base
suma_pares = 0
for i in range(1, 11):
    if i % 2 == 0:
        suma_pares += i

print("Suma de los pares:", suma_pares)
```

```python
# Agregamos acumulador para impares
suma_pares = 0
suma_impares = 0
for i in range(1, 11):
    if i % 2 == 0:
        suma_pares += i
    else:
        suma_impares += i

# Representamos el estado final como un diccionario
estado = {"suma_pares": suma_pares, "suma_impares": suma_impares}
print("Estado final:", estado)
```

### Explicación del cambio de estado:

Dentro del bucle, el estado (suma_pares y suma_impares) cambia en cada iteración según el valor de i. El flujo de control determina qué acumulador se actualiza.

## Parte 5: Programa de inventario con validación

```python
# Programa de inventario con validación
productos = []
while True:
    nombre = input("Ingrese producto (o 'fin' para salir): ")
    if nombre == "fin":
        break
    precio = float(input("Precio: "))
    cantidad = int(input("Cantidad: "))
    
    # Usamos continue para saltar productos con precio negativo o cantidad menor a 0
    if precio < 0 or cantidad < 0:
        print("Precio o cantidad inválidos, se salta el producto")
        continue
    
    productos.append({"nombre": nombre, "precio": precio, "cantidad": cantidad})

# Mostrar total
total = 0
for p in productos:
    total += p["precio"] * p["cantidad"]

print("Total del inventario:", total)
```

```python
# Mostrar el producto más caro usando max()
if productos:
    producto_mas_caro = max(productos, key=lambda p: p["precio"])
    print(f"Producto más caro: {producto_mas_caro}")
else:
    print("No hay productos")
```

```python
# Representar el estado final del inventario como lista de diccionarios
print("Estado final del inventario:", productos)
```
