# Ejercicio 6 - Sorted

```python
# a. Ordenar una lista de palabras alfabeticamente
palabras = ["zebra", "manzana", "banana", "uva", "pera"]
ordenadas = sorted(palabras)
print(ordenadas)
```

```python
# b. Ordenar una lista de tuplas por el primer elemento
tuplas = [(3, "tres"), (1, "uno"), (2, "dos"), (5, "cinco")]
ordenadas = sorted(tuplas, key=lambda x: x[0])
print(ordenadas)
```

```python
# c. Dada una lista de diccionarios, ordenar por precio (mayor a menor), y luego por nombre
productos = [
    ("Camisa", 30),
    ("Pantalón", 55),
    ("Medias", 10),
    ("Chaqueta", 80)
]
ordenados = sorted(productos, key=lambda x: (-x[1], x[0]))
print(ordenados)
```

```python
# d. Dada una lista de palabras, ordenarlas según su longitud
palabras = ["elefante", "gato", "hipopotamo", "perro", "raton"]
ordenadas = sorted(palabras, key=lambda x: len(x))
print(ordenadas)
```

```python
# e. Dada una lista de numeros, ordenarlos de tal forma que aparezcan primero los pares y luego los impares
numeros = [5, 2, 8, 1, 4, 3, 6, 7]
ordenados = sorted(numeros, key=lambda x: x % 2)
print(ordenados)
```

```python
# f. Convertir una palabra a un diccionario que cuente la aparicion de cada letra, ordenar dicho diccionario según la frecuencia de aparicion
from collections import Counter

palabra = "programacion"
contador = Counter(palabra)
ordenado = sorted(contador.items(), key=lambda x: x[1], reverse=True)
print(ordenado)
```

```python
# g. Dada una lista de tuplas, ordenar primero por el departamento del empleado y luego por la edad
empleados = [
    ("Maria", "Ventas", 30),
    ("Luis", "Tecnología", 25),
    ("Ana", "Ventas", 25),
    ("Pedro", "Tecnología", 28)
]
ordenados = sorted(empleados, key=lambda x: (x[1], x[2]))
print(ordenados)
```
