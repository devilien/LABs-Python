# Ejercicio 7 - Minimax (Juego NIM)

```python
import math

def minimax(n_piedras, es_max, profundidad=0):
    """
    Algoritmo minimax para el juego NIM.
    - MAX = Computadora (Inteligente)
    - MIN = Humano (Usuario)
    
    El jugador que toma la última piedra gana.
    """
    # Caso base: no quedan piedras
    if n_piedras == 0:
        # Si es MAX (computadora) y no quedan piedras, MAX pierde
        # Si es MIN (humano) y no quedan piedras, MIN pierde
        return -1 if es_max else 1
    
    if es_max:
        mejor_valor = -math.inf
        for tomar in [1, 2, 3]:
            if n_piedras >= tomar:
                valor = minimax(n_piedras - tomar, False, profundidad + 1)
                mejor_valor = max(mejor_valor, valor)
        return mejor_valor
    else:
        mejor_valor = math.inf
        for tomar in [1, 2, 3]:
            if n_piedras >= tomar:
                valor = minimax(n_piedras - tomar, True, profundidad + 1)
                mejor_valor = min(mejor_valor, valor)
        return mejor_valor

def obtener_mejor_movimiento(n_piedras):
    """Obtiene el mejor movimiento para la computadora."""
    mejor_movimiento = None
    mejor_valor = -math.inf
    
    for tomar in [1, 2, 3]:
        if n_piedras >= tomar:
            valor = minimax(n_piedras - tomar, False)
            if valor > mejor_valor:
                mejor_valor = valor
                mejor_movimiento = tomar
    
    return mejor_movimiento

def jugar_nim():
    """Juego NIM con algoritmo minimax."""
    n_piedras = 10  # Puedes cambiar el número inicial de piedras
    print(f"Juego NIM - Hay {n_piedras} piedras")
    print("En cada turno puedes tomar 1, 2 o 3 piedras")
    print("Pierde quien no puede mover (no quedan piedras)")
    print("El primer turno es del HUMANO (MIN)")
    print("-" * 40)
    
    turno_humano = True  # MIN empieza primero
    
    while n_piedras > 0:
        if turno_humano:
            print(f"\nPiedras restantes: {n_piedras}")
            try:
                tomar = int(input("Tu turno (1-3): "))
                if tomar not in [1, 2, 3] or tomar > n_piedras:
                    print("Movimiento inválido. Intenta de nuevo.")
                    continue
            except ValueError:
                print("Entrada inválida. Intenta de nuevo.")
                continue
        else:
            tomar = obtener_mejor_movimiento(n_piedras)
            print(f"\nComputadora toma {tomar} piedras")
        
        n_piedras -= tomar
        turno_humano = not turno_humano
    
    # Determinar ganador
    if turno_humano:  # El último que jugó fue la computadora
        print("\n¡Gana la Computadora (MAX)!")
    else:
        print("\n¡Gana el Humano (MIN)!")

# Para jugar, descomenta la siguiente línea:
# jugar_nim()
```
