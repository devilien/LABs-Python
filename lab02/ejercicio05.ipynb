# Ejercicio 5 - Reduce

```python
from functools import reduce

# a. Dada una lista de numeros, calcular la multiplicación de estos
numeros = [1, 2, 3, 4, 5]
multiplicacion = reduce(lambda x, y: x * y, numeros)
print(multiplicacion)
```

```python
# b. Dada una lista de palabras, concatenarlas en una sola
palabras = ["Hola", " ", "mundo", " ", "Python"]
concatenadas = reduce(lambda x, y: x + y, palabras)
print(concatenadas)
```

```python
# c. Dada una lista, hallar el mayor de todos
numeros = [5, 12, 3, 20, 8, 15]
mayor = reduce(lambda x, y: x if x > y else y, numeros)
print(mayor)
```
