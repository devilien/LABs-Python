# Ejercicio 1 - Funciones lambdas

## Parte 1: Definir lambdas básicas

```python
# a. Lambda que reciba un número y devuelva True si es múltiplo de 3, False en caso contrario
multiplo_de_3 = lambda x: x % 3 == 0
print(multiplo_de_3(9))  # True
print(multiplo_de_3(10))  # False
```

```python
# b. Lambda que reciba un número y devuelva su cubo
cubo = lambda x: x ** 3
print(cubo(3))  # 27
print(cubo(5))  # 125
```

```python
# c. Lambda que reciba dos números y devuelva su producto
producto = lambda x, y: x * y
print(producto(4, 5))  # 20
print(producto(7, 3))  # 21
```

```python
# d. Lambda que reciba dos números y devuelva el mayor
mayor = lambda x, y: x if x > y else y
print(mayor(10, 5))  # 10
print(mayor(3, 8))  # 8
```

```python
# e. Lambda que reciba una palabra y devuelva True si empieza con la letra "A" (o "a")
empieza_con_a = lambda palabra: palabra[0].lower() == 'a'
print(empieza_con_a("Amor"))  # True
print(empieza_con_a("casa"))  # False
```

```python
# f. Lambda que reciba una temperatura en Celsius y la convierta a Fahrenheit
celsius_a_fahrenheit = lambda c: (c * 9/5) + 32
print(celsius_a_fahrenheit(0))  # 32
print(celsius_a_fahrenheit(100))  # 212
```

## Parte 2: Lista de lambdas

```python
# Crear una lista que contenga tres lambdas:
# - Una que duplique un número
# - Una que le sume 10
# - Una que calcule su cuadrado

operaciones = [
    lambda x: x * 2,      # duplicar
    lambda x: x + 10,     # sumar 10
    lambda x: x ** 2      # cuadrado
]

# Pedir un número al usuario y aplicar cada lambda
numero = int(input("Ingrese un número: "))
for operacion in operaciones:
    print(operacion(numero))
```
