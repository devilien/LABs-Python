# Ejercicio 4 - Filter

```python
# a. Dada una lista de palabras, filtrar solo aquellas con mas de 4 letras
palabras = ["casa", "perro", "gato", "elefante", "raton"]
mas_de_4 = list(filter(lambda p: len(p) > 4, palabras))
print(mas_de_4)
```

```python
# b. Dada una lista filtrar los elementos que son nulos (None)
lista = [1, None, "hola", None, 5, None, "mundo"]
no_nulos = list(filter(lambda x: x is not None, lista))
print(no_nulos)
```

```python
# c. Dada una lista de palabras, filtrar aquellas que empiezan por una vocal
palabras = ["amor", "casa", "elefante", "gato", "uva", "perro"]
vocales = "aeiou"
empiezan_vocal = list(filter(lambda p: p[0].lower() in vocales, palabras))
print(empiezan_vocal)
```

```python
# d. Dada una lista de palabras, filtrar aquellas que son palindromos
palabras = ["ana", "casa", "reconocer", "gato", "oso", "perro"]
palindromos = list(filter(lambda p: p == p[::-1], palabras))
print(palindromos)
```

```python
# e. Dada una lista de numeros, filtrar los que terminan en 5
numeros = [15, 23, 35, 42, 55, 67, 75]
terminan_5 = list(filter(lambda n: n % 10 == 5, numeros))
print(terminan_5)
```

```python
# f. Dada una lista de diccionarios, filtrar los productos con precio mayor a 100
productos = [
    {"nombre": "Teclado", "precio": 80},
    {"nombre": "Mouse", "precio": 40},
    {"nombre": "Monitor", "precio": 300},
    {"nombre": "Webcam", "precio": 150}
]
caros = list(filter(lambda p: p["precio"] > 100, productos))
print(caros)
```
