# Ejercicio 3 - Map

```python
# a. Dada una lista de temperaturas en Celsius, convertirlas a Fahrenheit
celsius = [0, 10, 20, 30, 40]
fahrenheit = list(map(lambda c: (c * 9/5) + 32, celsius))
print(fahrenheit)
```

```python
# b. Dadas dos listas con la misma cantidad de elementos, generar una nueva sumando cada elemento
lista1 = [1, 2, 3, 4]
lista2 = [10, 20, 30, 40]
suma = list(map(lambda x, y: x + y, lista1, lista2))
print(suma)
```

```python
# c. Dada una lista de radios de circulos, generar la lista de sus areas
import math
radios = [1, 2, 3, 4, 5]
areas = list(map(lambda r: math.pi * r**2, radios))
print(areas)
```

```python
# d. Dada un diccionario, generar una lista de tuplas (nombre, valor+10%)
productos = {"pan": 1000, "leche": 2500, "café": 5000}
con_aumento = list(map(lambda item: (item[0], item[1] * 1.10), productos.items()))
print(con_aumento)
```

```python
# e. Dada una matriz, multiplicar todos los elementos por 10
matriz = [[1, 2, 3], [4, 5, 6], [7, 8, 9]]
multiplicada = [list(map(lambda x: x * 10, fila)) for fila in matriz]
print(multiplicada)
```
